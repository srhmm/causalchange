```python
import numpy as np
import pandas as pd

from causalchange import CMM
```

```python
rng = np.random.default_rng(0)
n = 300
z = rng.integers(0, 2, size=n)

x0 = rng.normal(size=n)
x1 = np.where(z == 0, 1.0 * x0, 2.0 * x0) + rng.normal(scale=0.2, size=n)
x2 = x1 + rng.normal(scale=0.2, size=n)

df = pd.DataFrame({"X0": x0, "X1": x1, "X2": x2})
df.head()
```

```python
model = CMM(
    mix_type="lin",
    k_max=2,
    seed=0,
)

model.fit(df)
```
